```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
x=np.linspace (0,20)
y=x

plt.plot(x,y,label='y=x')
plt.legend()
plt.grid(True)
plt.show
```

```python
x=np.linspace (0,20)
y=x**2

plt.plot(x,y,label='y=x^2')
plt.legend()
plt.grid(True)
plt.show
```

```python
x=np.linspace (0,20)
y=x**3

plt.plot(x,y,label='y=x^3')
plt.legend()
plt.grid(True)
plt.show
```
